## 3. Configurar ngrok

Pegue um token grátis em https://dashboard.ngrok.com/get-started/your-authtoken e cole abaixo.

```python
from pyngrok import ngrok, conf
import getpass

token = getpass.getpass('Cole seu ngrok authtoken: ')
conf.get_default().auth_token = token
tunnel = ngrok.connect(8000, bind_tls=True)
print('URL pública:', tunnel.public_url)
print('Cole essa URL no campo "URL do backend" do frontend.')
```

## 4. Subir o backend FastAPI em background (carrega TRIBE v2 na primeira request, ~1GB de pesos)

```python
import nest_asyncio, uvicorn, threading
nest_asyncio.apply()

config = uvicorn.Config('backend.main:app', host='0.0.0.0', port=8000, log_level='info')
server = uvicorn.Server(config)
threading.Thread(target=server.run, daemon=True).start()
print('Servidor rodando em background na porta 8000')
```

Abra a URL do ngrok no navegador → você verá o frontend servido pelo próprio backend.
